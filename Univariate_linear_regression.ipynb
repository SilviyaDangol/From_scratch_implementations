Here is a summary of some of the notation you will encounter.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

print(f"x_train = {x_train}")
print(f"y_train = {y_train}")
# Using floats instead of int because ML libraries generally owrk with floating point numbers
```

```python
# m is the number of  training examples
# x_train.shape -> (m,n) where m is number of exampels and n is nubmer of features

m= x_train.shape[0]
print(f"Number of training examples is: {m}")

```

```python
# Checking the training example(x_i,y_i)
for i in range(len(x_train)):
  x_i = x_train[i]
  y_i = y_train[i]
  print(f"(x^({i}), y^({i})=({x_i},{y_i})")
```

```python
# Plot the data points
plt.scatter(x_train, y_train, marker='o', c='b')
# Set the title
plt.title("House pricing")
# set the y_axis label
plt.ylabel ('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
# plt.show()
```

# Implementing model


```python
w =200
b= 100
print(f"w: {w}")
print(f"b: {b}")
```

```python
def compute_model_output(x,w,b):
  """
  Computes the prediction of a linear model
  Args:
    x(ndrray (m,)): Data, m examples
    w,b(scalar) : model parameters
  Returns
    f_wb(ndarray (m,)): model prediction
  """
  m = x.shape[0]
  f_wb = np.zeros(m)

  for i in range(m):
    f_wb[i] = w*x[i] + b

  return f_wb


```

```python
tmp_f_wb = compute_model_output(x_train, w , b)
print(tmp_f_wb) # y-pred
print(y_train) # target

# Plot our model prediction
plt.plot(x_train, tmp_f_wb, c='r', label='Our Prediction')

# Plot the data points
plt.scatter(x_train, y_train, marker='x', c='b', label='Actual Values')

# Set the title
plt.title("Housing Proces")

plt.ylabel('Price in 1000s of dollars')
plt.xlabel("size(1000 sqft)")

plt.legend()
plt.show()


```
