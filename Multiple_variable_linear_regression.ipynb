with multiple weights for one model( multiple linear regression)
each weight gets its own gradietn, and that gradient is an average over all training examples



For each wj: sum its gradient contribution across all m examples -> divide by m -> updatte wj.

different training example-> different contributions -> sum -> divide by m-> one gradient fofr one faeature.

We do this independently for w1, w2, ... wn for n features. and then all the features and b are updated simultaneously

```python
X_train = np.array([[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]]) # 4 features in each training example
y_train = np.array([460, 232, 178]) # three training samples so 3 target values
```

Parameter vector w, b
w is a vector witih n elemetns
Each element containins the parameter associated with one feature
b is a scalar parameter
w is aa 1-D NumPy vector

```python
b_init = 785.1811367994083
w_init = np.array([ 0.39133535, 18.75376741, -53.36032453, -26.42131618])
print(f"w_init shape: {w_init.shape}, b_init type: {type(b_init)}")
```

```python
# Single Prediction element by element

def predict_loop(x,w,b):
  n = x.shape[0]
  p = 0
  for i in range (n): # for each feature with its own weight
    p_i = x[i] * w[i]
    p += p_i
  p = p +b
  return p
```

```python

# single prediction vector

def predict(w,x,b):
  p = np.dot(w,x) + b
  return p
```

```python
# get a row of the training data
x_vec = X_train[0,:]
print(f"s_vec shape {x_vec.shape}, x_vect value: {x_vec}")

# make prediction with loop
f_wb = predict_loop(x_vec, w_init, b_init)
print(f"f_wb shape {f_wb.shape}, prediction{f_wb}")

# make prediction with vector:
f_wb_v = predict(x_vec, w_init, b_init)
print(f"f_wb_v shape{f_wb_v.shape}, prediction{f_wb_v}")
```

Cost with multiple varaibles

```python
def compute_cost(x, y, w,b):
  m = x.shape[0]
  cost_sum = 0.0
  for i in range(m):
    y_pred = np.dot(x[i],w) + b
    cost = (y_pred - y[i]) ** 2
    cost_sum +=  cost
  total_cost = cost_sum/(2*m)
  return total_cost

```

```python
# compute and display cost using pre chosen parameters
cost = compute_cost(X_train, y_train, w_init, b_init)
print(f"Cost at optimal w : {cost}")
```

Compute gradient with multiple variables


```python
def compute_gradient(X, y, w, b):
  m, n = X.shape
  dj_dw = np.zeros((n,)) # Initialized as 1D array to match w_init shape
  dj_db = 0.

  for i in range(m):
    err = (np.dot(X[i], w) + b) - y[i]
    for j in range(n):
      dj_dw[j] = dj_dw[j] + err * X[i, j]
    dj_db = dj_db + err

  dj_dw = dj_dw / m
  dj_db = dj_db / m

  return dj_db, dj_dw
```

```python
tmp_dj_db, tmp_dj_dw = compute_gradient(X_train, y_train, w_init, b_init)
print(f'dj_db at initial w,b: {tmp_dj_db}')
print(f'dj_dw at initial w,b: \n {tmp_dj_dw}')
```

```python
def gradient_descent(X, y, w_in, b_in, cost_function, gradient_function, alpha, num_iters):
    """
    Performs batch gradient descent to learn w and b. Updates w and b by taking
    num_iters gradient steps with learning rate alpha

    Args:
      X (ndarray (m,n))   : Data, m examples with n features
      y (ndarray (m,))    : target values
      w_in (ndarray (n,)) : initial model parameters
      b_in (scalar)       : initial model parameter
      cost_function       : function to compute cost
      gradient_function   : function to compute the gradient
      alpha (float)       : Learning rate
      num_iters (int)     : number of iterations to run gradient descent

    Returns:
      w (ndarray (n,)) : Updated values of parameters
      b (scalar)       : Updated value of parameter
      """

    # An array to store cost J and w's at each iteration primarily for graphing later
    J_history = []
    w = copy.deepcopy(w_in)  #avoid modifying global w within function
    b = b_in

    for i in range(num_iters):

        # Calculate the gradient and update the parameters
        dj_db,dj_dw = gradient_function(X, y, w, b)   ##None

        # Update Parameters using w, b, alpha and gradient
        w = w - alpha * dj_dw               ##None
        b = b - alpha * dj_db               ##None

        # Save cost J at each iteration
        if i<100000:      # prevent resource exhaustion
            J_history.append( cost_function(X, y, w, b))

        # Print cost every at intervals 10 times or as many iterations if < 10
        if i% math.ceil(num_iters / 10) == 0:
            print(f"Iteration {i:4d}: Cost {J_history[-1]:8.2f}   ")

    return w, b, J_history #return final w,b and J history for graphing
```

```python
# initialize parameters
initial_w = np.zeros_like(w_init)
initial_b = 0.0
# some gradient descent settings
iterations = 1000
alpha = 5.0e-7
# run gradient descent
w_final, b_final, J_hist = gradient_descent(X_train, y_train, initial_w, initial_b,
                                                    compute_cost, compute_gradient,
                                                    alpha, iterations)
print(f"b,w found by gradient descent: {b_final:0.2f},{w_final} ")
m,_ = X_train.shape
for i in range(m):
    print(f"prediction: {np.dot(X_train[i], w_final) + b_final:0.2f}, target value: {y_train[i]}")
```
